```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ------------------------------------------------------------
# ÉTAPE 1 : réparer le fichier
# ------------------------------------------------------------
# Dans le fichier d'origine, chaque ligne est entourée de guillemets.
# On enlève ces guillemets et on enregistre une copie propre.

fichier_origine = r"C:\SMART CITY DATA\02-06112026\DANDEGLA_EMMANUEL_RESET01\04_Base_de_donnees\nettoye\042donnees_SIMULEES_nettoyees.csv"
fichier_propre = r"C:\SMART CITY DATA\02-06112026\DANDEGLA_EMMANUEL_RESET01\04_Base_de_donnees\nettoye\donnees_propres.csv"

entree = open(fichier_origine, encoding="utf-8-sig")
sortie = open(fichier_propre, "w", encoding="utf-8")

for ligne in entree:
    ligne = ligne.strip()                     # enlève le retour à la ligne
    if ligne.startswith('"'):                 # si la ligne commence par "
        ligne = ligne[1:-1]                   # enlève le 1er et le dernier "
        ligne = ligne.replace('""', '"')      # remplace "" par "
    sortie.write(ligne + "\n")

entree.close()
sortie.close()
```

```python
# ------------------------------------------------------------
# ÉTAPE 2 : lire les données
# ------------------------------------------------------------

donnees = pd.read_csv(fichier_propre)

print("Nombre de répondants :", len(donnees))
print(donnees.head())
```

```python
# ------------------------------------------------------------
# Q1. Moyens de transport les plus utilisés
# ------------------------------------------------------------

print("\n----- Q1. Moyens de transport -----")
transport = donnees["moyen_transport"].value_counts()
print(transport)
print("En pourcentage :")
print(donnees["moyen_transport"].value_counts(normalize=True).round(3) * 100)

plt.figure(figsize=(8, 5))
plt.barh(transport.index, transport.values)
plt.title("Moyen de transport principal")
plt.xlabel("Nombre de répondants")
plt.tight_layout()
plt.savefig("Q1_transport.png")
plt.close()
```

```python
# ------------------------------------------------------------
# Q2. Motifs de déplacement
# ------------------------------------------------------------

print("\n----- Q2. Motifs de déplacement -----")
motifs = donnees["motif_deplacement"].value_counts()
print(motifs)

plt.figure(figsize=(8, 5))
plt.barh(motifs.index, motifs.values)
plt.title("Motif de déplacement")
plt.xlabel("Nombre de répondants")
plt.tight_layout()
plt.savefig("Q2_motifs.png")
plt.close()
```

```python
# ------------------------------------------------------------
# Q3. Zones de départ et de destination
# ------------------------------------------------------------

print("\n----- Q3. Zones de départ et d'arrivée -----")
print("Communes de départ :")
print(donnees["commune_depart"].value_counts())
print("\nQuartiers de départ :")
print(donnees["quartier_depart"].value_counts())
print("\nQuartiers d'arrivée :")
print(donnees["quartier_arrivee"].value_counts())
```

```python
# ------------------------------------------------------------
# Q4. Durée selon le moyen de transport et l'heure de départ
# ------------------------------------------------------------

print("\n----- Q4. Durée des trajets (minutes) -----")
print("Durée médiane de tous les trajets :", donnees["duree_trajet_min"].median())

print("\nDurée médiane selon le moyen de transport :")
print(donnees.groupby("moyen_transport")["duree_trajet_min"].median())

print("\nDurée médiane selon la période (pointe / hors pointe) :")
print(donnees.groupby("periode_depart")["duree_trajet_min"].median())

duree_mode = donnees.groupby("moyen_transport")["duree_trajet_min"].median()
plt.figure(figsize=(8, 5))
plt.barh(duree_mode.index, duree_mode.values)
plt.title("Durée médiane du trajet selon le moyen de transport")
plt.xlabel("Minutes")
plt.tight_layout()
plt.savefig("Q4_duree.png")
plt.close()
```

```python
# ------------------------------------------------------------
# Q5. Coût d'un aller simple selon le moyen de transport
# ------------------------------------------------------------

print("\n----- Q5. Coût d'un aller simple (FCFA) -----")
print("Coût médian :", donnees["cout_transport_fcfa"].median())
print("Nombre de coûts non renseignés :", donnees["cout_transport_fcfa"].isna().sum())

print("\nCoût médian selon le moyen de transport :")
print(donnees.groupby("moyen_transport")["cout_transport_fcfa"].median())
print("\nNombre de coûts renseignés par moyen de transport :")
print(donnees.groupby("moyen_transport")["cout_transport_fcfa"].count())
```

```python
# ------------------------------------------------------------
# Q6. Difficultés rapportées
# ------------------------------------------------------------

print("\n----- Q6. Difficultés -----")
difficultes = donnees["difficulte_principale"].value_counts()
print(difficultes)

plt.figure(figsize=(8, 5))
plt.barh(difficultes.index, difficultes.values)
plt.title("Principale difficulté rencontrée")
plt.xlabel("Nombre de répondants")
plt.tight_layout()
plt.savefig("Q6_difficultes.png")
plt.close()
```

```python
# ------------------------------------------------------------
# HYPOTHÈSES
# ------------------------------------------------------------

print("\n----- HYPOTHÈSES -----")

# H1 : durée médiane plus longue en heure de pointe
mediane_pointe = donnees[donnees["periode_depart"] == "Pointe"]["duree_trajet_min"].median()
mediane_hors = donnees[donnees["periode_depart"] == "Hors pointe"]["duree_trajet_min"].median()
nb_pointe = donnees[donnees["periode_depart"] == "Pointe"]["duree_trajet_min"].count()
nb_hors = donnees[donnees["periode_depart"] == "Hors pointe"]["duree_trajet_min"].count()

print("\nH1 : médiane pointe =", mediane_pointe, "min (", nb_pointe, "trajets )")
print("     médiane hors pointe =", mediane_hors, "min (", nb_hors, "trajets )")

if nb_pointe < 5 or nb_hors < 5:
    print("     Pas de conclusion : un groupe a moins de 5 trajets.")
elif mediane_pointe > mediane_hors:
    print("     Les données vont dans le sens de H1.")
else:
    print("     Les données ne vont pas dans le sens de H1.")

# H2 : le zémidjan est le moyen le plus cité
premier_transport = transport.index[0]
print("\nH2 : moyen le plus cité =", premier_transport)

if transport.values[0] == transport.values[1]:
    print("     Pas de conclusion : égalité en tête.")
elif premier_transport.startswith("Zémidjan"):
    print("     Les données vont dans le sens de H2.")
else:
    print("     Les données ne vont pas dans le sens de H2.")

# H3 : les embouteillages sont la difficulté la plus citée
premiere_difficulte = difficultes.index[0]
print("\nH3 : difficulté la plus citée =", premiere_difficulte)

if difficultes.values[0] == difficultes.values[1]:
    print("     Pas de conclusion : égalité en tête.")
elif premiere_difficulte == "Embouteillages":
    print("     Les données vont dans le sens de H3.")
else:
    print("     Les données ne vont pas dans le sens de H3.")
```
